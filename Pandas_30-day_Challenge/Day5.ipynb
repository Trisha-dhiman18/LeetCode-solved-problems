## Invalid Tweets

**Step 1: Import pandas**

```python
import pandas as pd
```

**Step 2: Create data**

```python
data = {
    "tweet_id": [1, 2],
    "content": ["Let us Code", "More than fifteen chars are here!"]
}
```

**Step 3: Create DataFrame**

```python
tweets = pd.DataFrame(data)
```

**Step 4: Find invalid tweets**

```python
filtered = tweets[tweets["content"].str.len() > 15]
```

**Step 5: Select required column**

```python
result = filtered[["tweet_id"]]
```

**Step 6: Print result**

```python
print(result)
```
