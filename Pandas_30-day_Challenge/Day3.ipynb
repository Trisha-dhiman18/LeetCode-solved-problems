**Step 1: Import pandas**

```python
import pandas as pd
```

**Step 2: Create Customers table**

```python
customers_data = {
    "id": [1,2,3,4],
    "name": ["Joe","Henry","Sam","Max"]
}

customers = pd.DataFrame(customers_data)
```

**Step 3: Create Orders table**

```python
orders_data = {
    "id": [1,2],
    "customerId": [3,1]
}

orders = pd.DataFrame(orders_data)
```

**Step 4: Find customers who ordered**

```python
orders["customerId"]
```

**Step 5: Filter customers who are NOT in orders**

```python
filtered_customers = customers[
    ~customers["id"].isin(orders["customerId"])
]
```

**Step 6: Select required column**

```python
result = filtered_customers[["name"]]
```

**Step 7: Rename column**

```python
result = result.rename(columns={"name":"Customers"})
```

```python
import pandas as pd

def find_customers(customers: pd.DataFrame, orders: pd.DataFrame) -> pd.DataFrame:
    
    result = customers[~customers['id'].isin(orders['customerId'])]
    
    return result[['name']].rename(columns={'name': 'Customers'})
```
