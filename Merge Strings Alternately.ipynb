## Merge Strings Alternately

**Step 1: Define the strings**

```python
word1 = "abc"
word2 = "pqr"
```

**Step 2: Create an empty result string**

```python
result = ""    # This will store the merged string.
```

**Step 3: Find minimum length**

```python
min_length = min(len(word1), len(word2))
```

**Step 4: Loop through characters**

```python
for i in range(min_length):
    result += word1[i]    #Add characters alternately
    result += word2[i]
result += word1[min_length:]       #Add remaining characters
result += word2[min_length:]
```

**LeetCode Style**

```python
def mergeAlternately(word1: str, word2: str) -> str:
    result = ""
    
    for i in range(min(len(word1), len(word2))):
        result += word1[i] + word2[i]
        
    result += word1[len(word2):] + word2[len(word1):]
    
    return result
```
